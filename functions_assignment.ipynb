##6. Explain the concept of generators in Python and how they are defined.
Generator functions in Python are a way to create iterators in a more concise and memory-efficient manner . These functions allow you to iterate over a potentially large sequence of data without loading the entire
sequence into memory.

```python
def square_number(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result
square_number(10)
```

##7.What are the advantages of using generators over regulor functions?

1.**Memory Efficiency**: Generators are memory-efficient because they don't store all values in memory at once.

```python
def list_nums():
    return [i for i in range(10**6)]  # This holds all 1 million numbers in memory
```

 2. **Lazy Evaluation (On-Demand Computation)**:
Generators compute values only when needed, which can improve performance.

```python
gen = (x * x for x in range(10))
```

3.**Use of yield**: Instead of using return to send a value back to the caller, a generator function uses ‘yield’.
When the generator function encounters a ‘yield’ statement, it returns the value specified in the yield and
pauses its execution state. The next time the generator is called, it resumes from where it left off.

##10. What is the difference between map()', 'reduce(), and 'fiitar() functions in Python?


 1. map() – Transforms items
Applies a function to each element of an iterable.

```python
nums = [1, 2, 3, 4]
squared = map(lambda x: x**2, nums)
print(list(squared))
```

 2. filter() – Filters items
Applies a function that returns True or False, and keeps only the True results.


```python
nums = [1, 2, 3, 4]
even = filter(lambda x: x % 2 == 0, nums)
print(list(even))
```

##4.Write a Python function that checks if a given number is prime or not from 1 to 200.



```python
def is_prime(n):
    if n < 2 or n > 200:
        return False  # Not prime if less than 2 or greater than 200
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
print(is_prime(5))
```

```python
print(is_prime(12))
```

##5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.


```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return self.a
```

```python
fib = FibonacciIterator(10)
for num in fib:
    print(num, end=" ")
```

##6. Write a generator function in Python that yields the powers of 2 up to a given exponent


```python
def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp
```

```python
for power in powers_of_two(5):
    print(power)
```

##7. Implement a generator function that reads a file line by line and yields each line as a string.


```python
def read_lines(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()

```

```python
file_name = 'example.txt'

for line in read_lines(file_name):
    print(line)
```

##10.Create a Python program that uses 'filter()" to remove all the vowels from a given string.


```python
def remove_vowels(input_string):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda x: x not in vowels, input_string))
input_string = "Hello, World!"
output_string = remove_vowels(input_string)
print("String without vowels:", output_string)
```
